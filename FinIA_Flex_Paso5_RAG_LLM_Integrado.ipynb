## Objetivo del notebook

Unir en un solo flujo automático los dos componentes construidos por separado:

- El **índice vectorial** (Paso 3), que permite buscar el fragmento de política relevante
  para un dato financiero dado.
- El **prompt maestro** (Paso 4), que redacta el reporte ejecutivo a partir de un dato
  financiero y un fragmento de contexto.

Hasta ahora, el fragmento de contexto se pasaba manualmente al prompt. En este notebook, la
búsqueda se vuelve automática: se le dan únicamente los datos financieros de un centro de
costo, y el sistema recupera el contexto relevante por sí mismo antes de generar el reporte.

Este flujo automático (recuperación + generación) es exactamente lo que define la técnica de
**RAG (Retrieval-Augmented Generation)**, y cumple de forma explícita el requisito 4 del caso
práctico.

## 4. Función de recuperación automática (Retrieval)

Esta función toma el centro de costo y la categoría de gasto de un caso, arma una consulta en
lenguaje natural, y busca en el índice vectorial los fragmentos más relevantes — sin
intervención manual. Esto reemplaza el paso donde antes copiabas y pegabas el contexto a
mano.

**Nota sobre una corrección aplicada:** en una primera versión, la consulta de búsqueda se
construía únicamente con la variación del mes actual (por ejemplo, "variación de 70.0%"), sin
incluir el histórico. Esto provocó que, en 2 de los 3 escenarios de prueba, el sistema no
recuperara los fragmentos correctos — por ejemplo, no encontró la regla del umbral de $50,000
MXN para Mantenimiento ni la regla de variación sostenida a 3 meses, a pesar de que ambas
reglas sí existen en el índice (se verificaron manualmente en el Paso 3). La causa fue que la
consulta no contenía las palabras clave necesarias para que la búsqueda semántica se acercara
al fragmento correcto. La corrección aplicada fue: (1) incluir el histórico completo en la
consulta, y (2) aumentar el número de fragmentos recuperados de `k=2` a `k=3`. Este hallazgo
es evidencia de que, en un sistema RAG, la calidad de la recuperación depende tanto del
índice como de qué tan bien se redacta la consulta de búsqueda — un punto valioso para el
informe de desarrollo.


```
def recuperar_contexto(centro_costo: str, categoria: str, resumen_situacion: str,
                        historico: str = "", k: int = 3) -> str:
    """
    Busca en el índice vectorial los fragmentos de política más relevantes
    para un caso financiero dado, y los devuelve como texto de contexto.

    Se incluye el histórico completo en la consulta (no solo la variación del mes
    actual), ya que reglas como la de "variación sostenida" o los umbrales de
    aprobación por categoría solo se recuperan de forma confiable si la consulta
    contiene las palabras clave asociadas (p. ej. "consecutivos", "aprobación").
    """
    consulta = (
        f"Política de aprobación y control de gasto para la categoría {categoria} "
        f"en el centro de costo {centro_costo}. Situación actual: {resumen_situacion}. "
        f"Histórico reciente: {historico}. "
        f"¿Qué umbral de aprobación o regla de variación sostenida aplica?"
    )
    resultados = vectorstore.similarity_search(consulta, k=k)

    fragmentos_texto = []
    for r in resultados:
        fuente = r.metadata["source"].split("/")[-1]
        fragmentos_texto.append(f'({fuente})\n"{r.page_content}"')

    return "\n\n".join(fragmentos_texto)


# Prueba rápida de la función de recuperación, de forma aislada
contexto_prueba = recuperar_contexto(
    centro_costo="Mantenimiento",
    categoria="Mantenimiento",
    resumen_situacion="gasto puntual muy por encima del presupuesto en un solo mes",
    historico="Pico aislado en julio, no observado en meses anteriores",
)
print(contexto_prueba)
```

## 4.5. Cálculo determinístico de indicadores (nueva sección — corrección numérica)

**Hallazgo que motiva esta sección:** en una prueba con los 3 escenarios, el modelo comparó
mal dos cifras — usó el gasto real total ($87,200 MXN) en vez de la variación sobre
presupuesto ($17,200 MXN) al evaluar si se excedía el umbral de $50,000 MXN de la categoría
Mantenimiento. Como consecuencia, citó la política equivocada (el umbral de aprobación) en
vez de la que realmente aplicaba (variación sostenida 3 meses consecutivos).

**Principio de la corrección:** los LLM no son calculadoras confiables. En lugar de pedirle al
modelo que compare montos y decida qué regla se activa, esos cálculos se hacen aquí en Python
— donde el resultado es siempre exacto — y se le entregan al modelo ya resueltos. El modelo
deja de "decidir" si algo excede un umbral; solo redacta con base en un resultado que ya viene
calculado. Esto no es un ajuste al prompt, es un cambio de responsabilidad: el cálculo pasa
del modelo (propenso a error) al código (determinístico).


```
UMBRALES_APROBACION_MXN = {
    "Materia Prima": 80000,
    "Mano de Obra Directa": 40000,
    "Energía": 30000,
    "Mantenimiento": 50000,
    "Logística/Fletes": 35000,
}


def clasificar_variacion(variacion_pct: float) -> str:
    """Clasifica la variación según los rangos definidos en POL-FIN-002, sección 3."""
    if variacion_pct >= 10:
        return "Variación significativa"
    elif variacion_pct >= 3:
        return "Variación moderada"
    elif variacion_pct >= -2.9:
        return "Dentro de rango normal"
    elif variacion_pct >= -10:
        return "Ahorro saludable"
    else:
        return "Ahorro atípico"


def calcular_racha_sobrecosto(variacion_pct_actual: float, historico_mensual: list) -> int:
    """
    Cuenta cuántos meses consecutivos de sobrecosto (variación positiva) hay,
    incluyendo el mes actual, recorriendo el histórico del más reciente al más antiguo.
    historico_mensual: lista de tuplas (mes, variacion_pct), en orden cronológico.
    """
    racha = 1 if variacion_pct_actual > 0 else 0
    if racha == 0:
        return 0
    for _, pct in reversed(historico_mensual):
        if pct > 0:
            racha += 1
        else:
            break
    return racha


def calcular_indicadores(categoria: str, variacion_mxn: float, variacion_pct: float,
                          historico_mensual: list = None) -> dict:
    """
    Calcula, en código (no en el LLM), los indicadores que antes el modelo debía
    inferir por sí mismo: clasificación de la variación, si excede el umbral de
    aprobación de su categoría, y si hay variación sostenida 3+ meses consecutivos.
    """
    historico_mensual = historico_mensual or []

    umbral = UMBRALES_APROBACION_MXN.get(categoria)
    excede_umbral = (umbral is not None) and (variacion_mxn > umbral)

    racha = calcular_racha_sobrecosto(variacion_pct, historico_mensual)
    variacion_sostenida = racha >= 3

    return {
        "clasificacion": clasificar_variacion(variacion_pct),
        "umbral_categoria_mxn": umbral,
        "excede_umbral": excede_umbral,
        "racha_meses_sobrecosto": racha,
        "variacion_sostenida": variacion_sostenida,
    }


# Prueba rápida con el caso del hallazgo (Escenario 1)
indicadores_prueba = calcular_indicadores(
    categoria="Mantenimiento",
    variacion_mxn=17200,
    variacion_pct=24.6,
    historico_mensual=[("Julio", 18.6), ("Agosto", 20.0)],
)
print(indicadores_prueba)
```

## 5. Función completa del flujo RAG + LLM

Esta es la función principal del copiloto: recibe los datos financieros de un caso, calcula
los indicadores de forma determinística (Sección 4.5), llama automáticamente a
`recuperar_contexto`, arma el prompt completo (system + few-shot + caso real, ya con los
indicadores calculados incluidos) y devuelve el reporte generado por el LLM.

A partir de aquí, usar el copiloto se reduce a llamar a esta única función con los datos de
cualquier centro de costo — ya no se requiere buscar contexto, ni calcular umbrales o rachas
a mano, ni confiar en que el modelo los calcule bien.


```
from groq import Groq

client = Groq()

def generar_reporte(centro_costo: str, categoria: str, presupuesto: float, real: float,
                     historico: str, historico_mensual: list = None,
                     responsable: str = None) -> str:
    """
    Flujo completo RAG + LLM: calcula indicadores de forma determinística,
    recupera contexto de políticas automáticamente, y genera el reporte ejecutivo.
    """
    variacion_mxn = real - presupuesto
    variacion_pct = variacion_mxn / presupuesto * 100

    indicadores = calcular_indicadores(categoria, variacion_mxn, variacion_pct, historico_mensual)

    resumen_situacion = f"variación de {variacion_pct:.1f}% ({variacion_mxn:,.0f} MXN)"
    contexto = recuperar_contexto(centro_costo, categoria, resumen_situacion, historico=historico)

    responsable_texto = responsable if responsable else "No especificado en los datos proporcionados"

    umbral_texto = (
        f"${indicadores['umbral_categoria_mxn']:,.0f} MXN" if indicadores["umbral_categoria_mxn"]
        else "sin umbral definido para esta categoría"
    )

    caso_entrada = f"""
DATOS:
Centro de costo: {centro_costo}
Categoría: {categoria}
Presupuesto: ${presupuesto:,.0f} MXN | Real: ${real:,.0f} MXN
Variación: {variacion_pct:.1f}% (${variacion_mxn:,.0f} MXN)
Histórico: {historico}
Responsable: {responsable_texto}

INDICADORES CALCULADOS POR EL SISTEMA (usa estos valores tal cual, no los recalcules ni los
contradigas — fueron calculados por código, no por ti):
- Clasificación de la variación: {indicadores['clasificacion']}
- Umbral de aprobación de la categoría "{categoria}": {umbral_texto}
- ¿La variación de este mes excede el umbral de su categoría?: {"SÍ" if indicadores['excede_umbral'] else "NO"}
- Meses consecutivos de sobrecosto (incluyendo el actual): {indicadores['racha_meses_sobrecosto']}
- ¿Aplica la regla de variación sostenida (3+ meses consecutivos de sobrecosto)?: {"SÍ" if indicadores['variacion_sostenida'] else "NO"}

CONTEXTO DE POLÍTICAS RECUPERADO (automático):
{contexto}
"""

    respuesta = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {"role": "system", "content": SYSTEM_PROMPT},
            {"role": "user", "content": EJEMPLO_FEW_SHOT_ENTRADA},
            {"role": "assistant", "content": EJEMPLO_FEW_SHOT_SALIDA},
            {"role": "user", "content": caso_entrada},
        ],
        temperature=0.3,
    )

    return respuesta.choices[0].message.content


print("Función generar_reporte() lista para usarse (con indicadores precalculados).")
```

## Resumen técnico (Paso 5)

**Proceso realizado:** integración del índice vectorial (Paso 3) con el prompt maestro
(Paso 4) en un flujo automático de dos funciones: `recuperar_contexto()` (retrieval) y
`generar_reporte()` (generación aumentada). El usuario final solo necesita proporcionar los
datos financieros de un centro de costo; la búsqueda de la política aplicable ya no requiere
intervención manual.

**Decisiones técnicas documentadas:**
- La consulta de búsqueda semántica se arma dinámicamente a partir del centro de costo, la
  categoría, la variación actual y el histórico completo — no solo el nombre del centro de
  costo — para mejorar la relevancia de los fragmentos recuperados.
- Se recuperan los 3 fragmentos más relevantes (`k=3`) por caso, balance entre dar suficiente
  contexto y no saturar el prompt con información irrelevante.
- Las reglas de grounding y anti-alucinación validadas en el Paso 4 se mantienen intactas en
  este flujo integrado.
- Los indicadores de clasificación, umbral y variación sostenida se calculan en Python
  (Sección 4.5) y se entregan ya resueltos al modelo, en lugar de pedirle que los calcule por
  sí mismo — el modelo solo redacta, no decide con base en aritmética propia.

**Hallazgos reales y correcciones aplicadas durante el desarrollo de este paso:**
1. *Fallo de recuperación:* en una primera corrida, el sistema no recuperó correctamente el
   contexto de política en 2 de los 3 casos, porque la consulta de búsqueda no incluía
   suficientes palabras clave. Corregido enriqueciendo la consulta con el histórico completo
   y aumentando `k` de 2 a 3.
2. *Error de razonamiento numérico:* en una corrida posterior, el modelo comparó el gasto real
   total contra el umbral de aprobación, en vez de la variación sobre presupuesto, citando la
   política equivocada. Corregido moviendo todo el cálculo de umbrales, clasificación y rachas
   de meses consecutivos a código Python (Sección 4.5), de modo que el modelo ya no realiza
   ninguna comparación numérica por su cuenta — solo redacta con base en indicadores ya
   calculados y verificados.

Ambos hallazgos, con sus respectivas correcciones, son evidencia de un proceso real de
control de calidad iterativo, y sustentan directamente el requisito 7 del caso práctico
(mecanismos de filtrado y control de calidad).

**Evidencia generada:** los 3 reportes de la Sección 6, junto con la verificación cruzada de
la Sección 7, constituyen la evidencia principal de que el sistema cumple el requisito 4 del
caso práctico (RAG) de forma funcional, con salidas numéricamente consistentes.

**Siguiente paso:** Paso 6 — fine-tuning ligero (LoRA) en un modelo pequeño, como
demostración adicional de personalización del LLM sobre el formato de reporte.
